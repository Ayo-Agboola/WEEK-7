```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib as plt
```

```python
insurance_df = pd.read_csv('insurance.csv')
insurance_df
```

BAR CHART

```python
#Seaborn 

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(5, 3))
sns.barplot(data = insurance_df, x='smoker', y='charges', estimator='mean', color='red')
plt.title('Bar chart for total bill by smoker')
```

```python
temp_df = insurance_df[['smoker', 'charges']].groupby(by='smoker').mean()
plt.figure(figsize=(7, 4))
plt.bar(temp_df.index, temp_df['charges'])
plt.title('Bar Chart of Total Bill vs SmokinG Habit')
```
