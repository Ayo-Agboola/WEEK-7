```python
insurance_df = insurance_df.astype(
    {'age': 'float', 'children': 'float'}
)

insurance_df.info()
```

```python
#fillna

median = insurance_df['age'].median()
insurance_df['age'].fillna(value=median, inplace=True)
```

```python
insurance_df.dropna(inplace=True)
```

```python
insurance_df.describe()
```

```python
insurance_df.select_dtypes(include='float').corr(method='spearman')
```

```python
# Basic Arithmetic

insurance_df['tax'] = insurance_df['charges'] * 0.05
insurance_df.head(n=5)
```

```python
# Create a new column based on condition

insurance_df['age_class'] = ['teenager' if age < 20 
                             else 
                             'adult' if age < 59 
                             else 
                             'old'
                             for age in insurance_df['age']]
insurance_df.head()
```

```python
#Aggregation and Grouping

insurance_df[['charges', 'region']].groupby(by='region', sort=True).sum()
```
