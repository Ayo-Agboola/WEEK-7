4. Conditional Statements

Conditional statements allow a program to make decisions based on certain conditions.

Python uses the keywords if, elif, and else:

if condition:
    # code block
elif another_condition:
    # code block
else:
    # code block
Conditions must evaluate to True or False.
Indentation (usually 4 spaces) defines the scope of each block.
Only one block runs per evaluation sequence.
Conditionals are fundamental for controlling program flow.

5.3. Dictionary

A dictionary stores data in key-value pairs inside curly braces {}. 

Keys must be unique and immutable (usually strings or numbers).

Unlike lists that use numbered positions, dictionaries let you look up values using descriptive labels (keys).

• Keys and Values. Each key must be unique (like "name" or "age"), and it points to an associated value (like "Alex" or 25).
• Fast lookups. You access, add, or update data directly by referencing the key inside square brackets.


Example:

student = {
    "name": "Alice",
    "age": 25
}
Common operations:

Access: student["name"]
Add or update: student["grade"] = "A"

```python
# Create a dictionary state and capital.
nigeria = {
    "Akwa-Ibom": "Uyo",
    "Oyo": "Ibadan",
    "Lagos": "Ikeja"
}

# Print the key and value.
print(nigeria)
print(nigeria["Oyo"]) # Print the capital of Oyo state
print(nigeria["Akwa-Ibom"]) # Print the capital of Akwa-Ibom

# Add a new state "Ogun" and update the capital. 
nigeria["Ogun"] = "Abeokuta"
print(nigeria["Ogun"]) # Print the capital of Ogun state
print(nigeria) # Print the updated dictionary

# Update
nigeria.update({"Ekiti": "Ado-Ekiti"})
print(nigeria) # Print the updated dictionary
```

```python
# Create a dictionary for a product with name, price and quantity.

product = {
    "name": "Laptop",
    "price": 100000,
    "quantity": 10
}

print(product) # Print the product dictionary

total_price = product["price"] * product["quantity"]
print(total_price) # Print the total price of the product
```

```python
# Write a program that accepts name of a person and their year
# of birth and create a dictionary of key name and age as the value.

# user input
from datetime import date


name = input("Enter your name: ")
year_of_birth = int(input("Enter your year of birth: "))

# Calculate age
current_year = date.today().year
age = current_year - year_of_birth

# Create a dictionary
person = {
    "name": name,
    "age": age
}

print(person) # Print the dictionary
```

2. For Loop

A for loop is used to iterate over a known sequence of items, such as a list, a string of text, or a range of numbers. It runs a block of code once for every item in that collection.
• Best used when: You know how many times you want the loop to run, or you want to touch every item in a list.
• Indentation matters: The code inside the loop must be indented (usually by 4 spaces) so Python knows what to repeat.


```python
# Create a list of numbers from 1 to 5
numbers = [1, 2, 3, 4, 5]
print(numbers) # Print the list
```

```python
# Use a for loop to print each number in the list.
for number in numbers:
    print(number) # Print each number
```

```python
# Use a for loop to print the square of each number in the list.
for number in numbers:
    print(f'the square of {number} is {number ** 2}') # Print the square of each number
```

```python
#Write a program that returns all the even numbers with a range.

lower_num = int(input("Enter the lower range: "))
upper_num = int(input("Enter the upper range: "))

numbers = list(range(lower_num, upper_num + 1)) # Create a list of numbers in the range

for number in numbers:
    if number % 2 == 0:
        print(f'{number} is even') # Print even numbers
```

3. While Loop

A while loop repeats a block of code as long as a specific condition remains True. Once the condition becomes False, the loop stops.
• Best used when: You do not know in advance how many times the code needs to repeat.
• Safety rule: Always include a way for the condition to eventually become False, otherwise you will create an infinite loop that freezes your program.


```python
# Create a variable with value 5

max_number = 10

# Use WHILE loop to print a variable as long as it is less than 5

number = 0

while number < max_number:
    print(number)
    number += 1
```

Functions

A function is a reusable block of code that performs a specific task. Instead of rewriting the same code multiple times, you define a function once with the def keyword and call it whenever needed.
• Parameters and Arguments. You can pass input values (arguments) into a function so it works dynamically with different data.
• Return values. A function can send a final result back to the main program using the return keyword.


```python
# Write a function that takes a number as input and return the output as square

def get_square(number):
    square = number ** 2
    return square
```

```python
# call the functions with different numbers and print the results. 

print(get_square(number=2))
print(get_square(10))
```

```python
# Write a function that takes two numbers and return their sum

def add_two_numbers(first_number, second_number):
    addition = first_number + second_number
    return addition

```
