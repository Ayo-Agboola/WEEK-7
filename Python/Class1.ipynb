ARITHMETIC OPERATIONS

OPERATORS      DESCRIPTION
+               ADDITION
-               SUBTRACTION
*               MULTIPLICATION
/               FLOAT DIVISION
//              FLOOR DIVISION (INTEGER DIVISION)
%               REMAINDER 
**              POWER

```python
# Variable Assignments

a = 2
b = 4

# Variables for Operations
addition = a + b
print(addition)

sub = a - b
print(sub)

multiply = a * b
print(multiply)

float_div = a / b
print(float_div)

floor_div = b // a
print(floor_div)

modulus = a % b
print(modulus)

power = a ** b
print(power)

# CALCULATE THE AREA OF A CIRCLE with radius r = 5
# (USE pie = 3.1416 and Formula: Area = pie * r ** 2)

pie = 3.1416
r = 5

area = pie * r ** 2

print(area)
```

COMPARISON AND LOGICAL OPERATIONS

Comparison operators compare two values and return a boolean result (True or False)

OPERATORS    MEANING
==           EQUAL TO
!=           NOT EQUAL TO
>            GREATER THAN
<            LESS THAN
>=           GREATER THAN OR EQUAL TO
<=           LESS THAN OR EQUAL TO
AND          TRUE IF BOTH ARE TRUE
OR           TRUE IF AT LEAST ONE IS TRUE
NOT          NEGATES A BOOLEAN VALUE

```python
# CREATE VARIABLES

x = 6
y = 36

equal_to = x == y
print(equal_to)

not_equal_to = x != y
print(not_equal_to)

less_than = x < y
print(less_than)

greater_than = x > y
print(greater_than)

less_than_or_equal_to = x <= y
print(less_than_or_equal_to)

greater_than_or_equal_to = x >= y
print(greater_than_or_equal_to)

# Logical Operators

is_single = False
is_happy = False

print(is_single and is_happy)
print(is_single or is_happy)
print(not is_single)
```
